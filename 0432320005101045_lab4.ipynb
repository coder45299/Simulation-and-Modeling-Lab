```python
import random  # to create random numbers

def estimate_pi(num_points):
    inside = 0  # how many points inside the circle

    for i in range(num_points):
        x = random.random()  # pick a random number from 0 to 1
        y = random.random()
        if x*x + y*y <= 1:
            inside += 1  # it's inside the circle

    pi = 4 * inside / num_points
    print("Estimated pi for", num_points, "points is:", pi)

# Try this with different number of points
estimate_pi(1000)
estimate_pi(5000)
estimate_pi(10000)
estimate_pi(50000)
```

```python
import random
import numpy as np
import math

# ==========================================
# Part 1 - Setup & Personalization
# ==========================================

# My details
student_id = "0432320005101050"
full_name = "Zakia Sultana Chadni"

# Extracting R from last 2 digits
R = int(student_id[-2:])
radius = R / 10

# Counting letters (without spaces)
letters = len(full_name.replace(" ", ""))
min_points = 200 * letters

print("Radius:", radius)
print("Base Points:", min_points)
print()
```

```python
# Part 2 - Simulation Logic
# ==========================================

def run_test(total_points, r):
    inside_count = 0
    half = total_points // 2

    # First half using random.uniform
    for i in range(half):
        x = random.uniform(0, r)
        y = random.uniform(0, r)
        if (x*2 + y*2) <= (r*2):
            inside_count = inside_count + 1

    # Second half using numpy.random.rand
    for i in range(total_points - half):
        x = np.random.rand() * r
        y = np.random.rand() * r
        if (x*2 + y*2) <= (r*2):
            inside_count = inside_count + 1

    return inside_count
```

```python
# Part 3 - Output & Analysis
# ==========================================

import math

print(f"{'Run':<6} {'Total':<10} {'Inside':<10} {'Pi':<10} {'Error':<10}")

for i in range(1, 6):
    pts = min_points * i
    inside = run_test(pts, radius)
    pi_val = 4 * (inside / pts)

    print(
        f"{i:<6} {pts:<10} {inside:<10} {pi_val:.4f}     {abs(math.pi - pi_val):.4f}"
    )
```
