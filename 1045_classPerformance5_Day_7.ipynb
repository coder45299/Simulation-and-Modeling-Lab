Chi-Square Goodness-of-Fit Test in Python

## Import Necessary Libraries

````python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
````

````python
##class performance 5

# 1. Student ID last 2 digit (d) and Observed Frequencies (O) calculation
d = 45

# observed frequencies:
# C: 20 + d = 20 + 45 = 65
# C++: 25 + (d % 10) = 25 + (45 % 10) = 25 + 5 = 30
# Java: 30 + (d % 7) = 30 + (45 % 7) = 30 + 3 = 33
# Python: 35 + (d % 5) = 35 + (45 % 5) = 35 + 0 = 35

observed = np.array([65, 30, 33, 35])
print(f"1. Student ID's last two digits (d): {d}")
print(f"2. Calculated Observed Frequencies (O): {observed}")

# 3 & 4. Null and Alternative Hypothesis
print("\n3. Null Hypothesis (H0): Students have an equal preference for all four programming languages.")
print("4. Alternative Hypothesis (H1): Students do not have an equal preference for the four programming languages.")

# 5. Expected Frequencies (E) assuming uniform distribution
total = np.sum(observed)
expected = np.full_like(observed, total / len(observed), dtype=float)
print(f"\n5. Expected Frequency (E) for each category: {expected}")

# 6. Calculated Chi-Square Statistic
chi_square_stat = np.sum(((observed - expected) ** 2) / expected)
print(f"\n6. Calculated Chi-Square statistic: {chi_square_stat:.2f}")

# 7. Degrees of Freedom
df = len(observed) - 1
print(f"7. Degrees of Freedom (df): {df}")

# 8. Critical Value or p-value
alpha = 0.05  # Significance level
critical_value = stats.chi2.ppf(1 - alpha, df)
print(f"8. Critical Value (at 5% significance level): {critical_value:.2f}")

# 9. Final Decision
print("\n9. Final Decision:")
if chi_square_stat > critical_value:
    print("Reject the null hypothesis. Observed data does not match the expected distribution.")
else:
    print("Fail to reject the null hypothesis. Observed data matches the expected distribution.")

# 10. Brief Conclusion
print("\n10. Conclusion:")
print("Since the chi-square statistic is greater than the critical value, we reject H0. This concludes that students do not have an equal preference for the four programming languages.")

# Plotting Observed vs Expected Frequencies
plt.bar(range(1, len(observed) + 1), observed, width=0.4, label='Observed', alpha=0.7)
plt.bar(np.arange(1, len(observed) + 1) + 0.4, expected, width=0.4, label='Expected', alpha=0.7)
plt.xlabel('Categories (1: C, 2: C++, 3: Java, 4: Python)')
plt.ylabel('Frequencies')
plt.title('Observed vs Expected Frequencies')
plt.legend()
plt.show()
````

### Chi-Square Goodness-of-Fit Test in Python

#### Objectives
- To perform a Chi-square goodness-of-fit test using Python to determine if a given dataset follows a hypothesized distribution.

#### Apparatus Required
- A computer with Python installed and libraries such as `scipy`, `numpy`, `matplotlib` installed. You can install the necessary libraries using:

#### Theory
The Chi-square goodness-of-fit test compares the observed frequencies to the expected frequencies and determines whether there is a significant difference between them. The test statistic is calculated as:
\[
\chi^2 = \sum \frac{(O_i - E_i)^2}{E_i}
\]
Where:
- \(O_i\) = observed frequency in category \(i\)
- \(E_i\) = expected frequency in category \(i\)

Degrees of freedom (df) is calculated as:
\[
df = \text{{number of categories}} - 1
\]
The result is compared with the critical value from the Chi-square distribution at a specified significance level (\(\alpha = 0.05\)).

#### Steps
1. Collect the observed data (frequencies).
2. Define the expected distribution (e.g., uniform, normal).
3. Compute the expected frequencies based on the distribution.
4. Calculate the Chi-square statistic.
5. Compare the statistic with the Chi-square distribution critical value to accept or reject the null hypothesis.


### Explanation of the Code
1. **Observed Data**: The observed frequencies are defined in the `observed` array. This can be replaced with your own dataset.
2. **Expected Frequencies**: The expected frequencies are calculated assuming a uniform distribution. For a uniform distribution, each category is expected to have an equal proportion of the total sum of the observed data.
3. **Chi-Square Statistic**: We use the `chisquare` function from `scipy.stats` to calculate the Chi-square statistic and the p-value.
4. **Degrees of Freedom**: The degrees of freedom is calculated as `df = len(observed) - 1`.
5. **Critical Value**: The critical value at the 0.05 significance level is obtained from the Chi-square distribution using `stats.chi2.ppf`.
6. **Conclusion**: We compare the Chi-square statistic with the critical value to determine whether to reject the null hypothesis.
7. **Visualization**: A bar chart is plotted to compare the observed and expected frequencies for each category.

### Output
The output will display:
- The Chi-square statistic
- The degrees of freedom
- The critical value
- The p-value
- A conclusion whether to reject or fail to reject the null hypothesis

It will also display a bar graph comparing the observed vs expected frequencies.

### Discussion and Conclusion
- **Chi-square statistic**: It quantifies the deviation between the observed and expected frequencies.
- **Degrees of Freedom (df)**: It represents the number of categories minus one. It affects the Chi-square distribution.
- **P-value**: The p-value indicates the probability of observing a statistic as extreme as the one calculated, given that the null hypothesis is true.
  - If the p-value is less than 0.05, reject the null hypothesis, indicating that the observed data does not fit the expected distribution.
  - If the p-value is greater than 0.05, fail to reject the null hypothesis, indicating that the observed data reasonably fits the expected distribution.

### VIVA Questions
1. **What is the purpose of the Chi-square goodness-of-fit test?**
   - The Chi-square goodness-of-fit test evaluates whether the observed frequencies of a dataset significantly differ from the expected frequencies under a specified hypothesis.
   
2. **How do you calculate the Chi-square statistic?**
   - The Chi-square statistic is calculated using the formula:
     \[
     \chi^2 = \sum \frac{(O_i - E_i)^2}{E_i}
     \]
     where \(O_i\) and \(E_i\) are the observed and expected frequencies, respectively.

3. **What is the significance of degrees of freedom in the Chi-square test?**
   - The degrees of freedom determine the shape of the Chi-square distribution. It is calculated as the number of categories minus one, and it influences the critical value and the p-value.

4. **What does a p-value represent in the Chi-square test?**
   - The p-value represents the probability of observing the test statistic or something more extreme, assuming the null hypothesis is true. A small p-value (less than 0.05) suggests rejecting the null hypothesis.
